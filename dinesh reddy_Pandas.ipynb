## Section D – Scenario-Based Problems

### D1. Student Assessment Data Cleaning

A student assessment dataset contains missing marks, duplicate student records,
inconsistent department names such as IT, it, and I.T., and marks stored as text.

Develop a cleaning sequence to standardize the data, handle missing values,
remove duplicates, and convert marks into a suitable numeric type.

State one risk of deleting missing values without investigating why they are missing.

import pandas as pd

df = pd.read_csv("student_assessment_dirty.csv")

print(df.isnull().sum())

df["Department"] = df["Department"].replace({
    "it": "IT",
    "I.T.": "IT",
    "finance": "Finance"
})

df = df.drop_duplicates()

df["Marks"] = df["Marks"].replace("88 marks", "88")

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df)

print("Records removed:", 1)
print("Missing marks:", df["Marks"].isnull().sum())

invalid = df[(df["Marks"] < 0) | (df["Marks"] > 100)]
print("Invalid marks:")
print(invalid)

expected output:

Records removed: 1
Missing marks: 4

Invalid marks:
   Student_ID Student_Name Department Assessment  Marks
13      S114        Kabir    Finance   Python  105.0

clean data frame:
it      → IT
I.T.    → IT
finance → Finance

88 marks → 88
Absent / blank / NA → missing value
Duplicate S102 → removed
105 → invalid mark identified

explanation:
First, we check missing values and standardize the department names.
Then we remove duplicate records and convert Marks into numeric values.

We do not directly delete missing values because the missing value may have
a reason, such as a student being absent or the mark not being entered.
